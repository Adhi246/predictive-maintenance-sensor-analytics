```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/train_FD001_descriptive.csv")

long_names = df.columns[5:26].tolist()
short_names = [f"s{i}" for i in range(1, 22)]
name_map = dict(zip(short_names, long_names))
df = df.rename(columns=dict(zip(long_names, short_names)))

sensors = short_names
print(df.shape)
pd.Series(name_map)
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Number of engines:", df["engine"].nunique())

# Check the RUL column is what we think it is
df["max_cycle"] = df.groupby("engine")["cycle"].transform("max")
print("RUL matches (max cycle - cycle):", (df["max_cycle"] - df["cycle"] == df["RUL"]).all())

print(df.groupby("engine")["cycle"].max().describe())
```

```python
stds = df[sensors].std()
constant = stds[stds == 0].index.tolist()
print("Constant sensors:", constant)
stds.sort_values().round(4)
```

```python
useful = [s for s in sensors if s not in constant]
corr = df[useful + ["RUL"]].corr()["RUL"].drop("RUL").sort_values()
result = pd.DataFrame({"correlation_with_RUL": corr.round(3),
                       "sensor_name": corr.index.map(name_map)})
result
```

```python
key = ["s11", "s4", "s12", "s7"]
e = df[df["engine"] == 1]

fig, axes = plt.subplots(2, 2, figsize=(12, 7), sharex=True)
for ax, s in zip(axes.ravel(), key):
    ax.plot(e["cycle"], e[s], linewidth=0.6, label="raw")
    ax.plot(e["cycle"], e[s].rolling(10).mean(), linewidth=1.8, label="10-cycle average")
    ax.set_title(f"{s}: {name_map[s]}", fontsize=9)
axes[0, 0].legend()
axes[1, 0].set_xlabel("Cycle")
axes[1, 1].set_xlabel("Cycle")
plt.suptitle("Engine 1: key sensors over its life (NASA C-MAPSS FD001, simulated data)")
plt.tight_layout()
plt.savefig("../images/engine1_key_sensors.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
def first_warning(g, s, direction, base_n=30, win=10, k=3):
    base = g[s].iloc[:base_n]               # healthy period: first 30 cycles
    mu, sd = base.mean(), base.std()
    smooth = g[s].rolling(win).mean()
    if direction == "up":
        flag = smooth > mu + k * sd
    else:
        flag = smooth < mu - k * sd
    flag = flag & (g["cycle"] > base_n)
    if flag.any():
        return g.loc[flag.idxmax(), "RUL"]  # remaining life at first warning
    return np.nan

direction = {"s11": "up", "s4": "up", "s12": "down", "s7": "down"}

rows = []
for s in key:
    ruls = np.array([first_warning(g, s, direction[s]) for _, g in df.groupby("engine")])
    warned = ruls[~np.isnan(ruls)]
    rows.append({
        "sensor": s,
        "engines_warned": len(warned),
        "never_warned": int(np.isnan(ruls).sum()),
        "median_cycles_left_at_warning": float(np.median(warned)),
        "10th_percentile_cycles_left": float(np.percentile(warned, 10)),
    })
pd.DataFrame(rows)
```

```python
rul_warn = np.array([first_warning(g, "s11", "up") for _, g in df.groupby("engine")])
life = df.groupby("engine")["cycle"].max().values

print("s11 earliest warning (most cycles left):", rul_warn.max())
print("s11 90th percentile of cycles left:", round(np.percentile(rul_warn, 90), 1))
print("s11 median share of life left at warning (%):", round(np.median(rul_warn / life) * 100, 1))

plt.figure(figsize=(7, 4))
plt.hist(rul_warn, bins=20)
plt.xlabel("Cycles left when first warning appeared")
plt.ylabel("Number of engines")
plt.title("s11 early warning lead time, 100 engines (simulated NASA data)")
plt.savefig("../images/s11_warning_lead_time.png", dpi=150, bbox_inches="tight")
plt.show()
```
